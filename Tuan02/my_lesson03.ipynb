#### 2. Các thao tác trên dãy số a

(b) Tìm số lượng các phần tử của dãy số <br/>
(c) In ra giá trị lớn nhất, giá trị nhỏ nhất và giá trị trung bình <br/>
(d) Tính phương sai của dãy số $\sigma^2 = \frac{\sum{\left(a_i - \mu\right)^2}}{N - 1}$ với $\mu = \frac{\sum{a_i}}{N}$ và độ lệch chuẩn $var = \sqrt{\sigma^2}$

```python
print("2. Các thao tác tính toán đơn giản với dãy số a")
a2 = a ** 2
na = a.size

# a. Bình phương các phần tử của a
print("a. Binh phuong a: ", a2)
# b. Độ dài của dãy a
print("\nb. Do dai a: ", na)

# c. Giá trị lớn nhất, nhỏ nhất, trung bình
v_max = np.max(a)
v_min = np.min(a)
v_avg = np.mean(a)

print(f'\nc. Gia tri lon nhat: [{v_max}], gia tri nho nhat: [{v_min}], gia tri trung binh [{v_avg}]')

# d. Phương sai và độ lệch chuẩn
v_sigma = np.var(a)
v_var = np.std(a)
print(f'\nd. Phuong sai: [{v_sigma: .2f}] va do lech chuan: [{v_var: .2f}]')
```

#### 3. Toán tử trên hai dãy số
(a) Tạo dãy số x có 10 phần tử từ 1 đến 10 (dùng hàm np.arange) <br/>
(b) Liệt kê và đếm số lượng phần tử chẵn và lẻ có trong dãy a và x <br/>
(c) Tính khoảng cách giữa các phần tử ở vị trí lẻ của mảng a và x với $s = \sqrt{\sum{\left(a_i - x_i\right)^2}}$ với $i$ lẻ<br/>
(d) Tìm khoảng cách nhỏ nhất giữa tập a và x. Ví dụ: min({1, 3, 5}, {2, 10, 8}} = 1

```python
print("3. Xử lý khác trên dãy số")
# a. Lấy các phần tử chẵn trong dãy a
x = a[a % 2 == 0]
print(f'a. Day so x: {x}')
```

```python
# Kiểm tra phần tử chẵn (True/False)
a_la_chan = a % 2 == 0
x_la_chan = x % 2 == 0

# Lấy các phần tử chẵn
a_chan = a[a_la_chan]
x_chan = x[x_la_chan]

# Đếm số phần tử chẵn
n_a_chan = a_chan.size
n_x_chan = x_chan.size
print( "b. Cac phan tu chan trong day a, b: ")
print(f"+ Day a: {a}")
print(f"  - Loc chan: {a_la_chan}")
print(f"  - Day chua phan tu chan trong a: {a_chan}, so luong {n_a_chan} phan tu.")
print(f"+ Day x: {x}")
print(f"  - Loc chan: {x_la_chan}")
print(f"  - Day chua phan tu chan trong a: {x_chan}, so luong {n_x_chan} phan tu.")
```

```python
# Vị trí lẻ trong mảng (index lẻ)
v_le = np.arange(1, min(len(a), len(x)), 2)

# Các phần tử ở vị trí lẻ
a_le = a[v_le]
x_le = x[v_le]

# Hiệu các phần tử ở vị trí lẻ
hieu_le = a_le - x_le 

# Khoảng cách (chuẩn Euclid)
kc_le = np.linalg.norm(hieu_le)

print( "c.Tính hiệu các phần tử ở vị trí lẻ của mảng a và x: ")
print(f"+ Vi tri le trong a va x: {v_le}")
print(f"+ Day so a: {a} va cac so o vi tri le: {a_le}")
print(f"+ Day so x: {x} va cac so o vi tri le: {x_le}")
print(f"+ Hieu cua a va x o vi tri le: {hieu_le}")
print(f"+ Khoang cach cac phan tu cua a va x o vi tri le: {kc_le: .2f}")
```

```python
print("d. Tìm khoảng cách nhỏ nhất giữa tập a và x. Ví dụ: min({1, 3, 5}, {2, 10, 8}} = 1")
"""
Gợi ý: Áp dụng toán tử broatcasting
[[1],                    [[1, 1, 1],        [[2, 10, 8],    =   [[-1, -9, -7],
 [3],   - [2, 10, 8] =    [3, 3, 3],   -     [2, 10, 8],         [ 1, -7, -5],  --> np.min(np.abs(...)) = 1
 [5]]                     [5, 5, 5]]         [2, 10, 8]]         [ 3, -5, -3]]
"""
# Tăng số chiều cho a để áp dụng broadcasting
diff = a[:, np.newaxis] - x

# Khoảng cách nhỏ nhất giữa hai tập
v_min = np.min(np.abs(diff))

print(f"+ Day a: {a}")
print(f"+ Day x: {x}")
print(f"+ Khoang cach nho nhat: {v_min}")
```

(b) Tính tổng, hiệu, tích của từng phần tử trên hai ma trận A và B <br/>
(c) Tính phép nhân ma trận Y = A x X

```python
# a. Tạo các ma trận
A = np.random.randint(1, 10, (4, 3))
B = np.random.randint(1, 10, (4, 3))
X = np.random.randint(1, 10, (3, 4))

# b. Tổng, hiệu, tích từng phần tử
AB_add = A + B
AB_sub = A - B
AB_mul = A * B   # nhân từng phần tử (Hadamard)

# c. Nhân ma trận
Y = A @ X        # hoặc np.dot(A, X)

print("a. Tạo hai ma trận A và B có 4 dòng x 3 cột và X có 3 dòng x 4 cột")
print(f"+ Ma tran A: \n{A}")
print(f"+ Ma tran B: \n{B}")
print(f"+ Ma tran X: \n{X}")

print("\nb. Tính tổng, hiệu, tích của từng phần tử trên hai ma trận A và B")
print(f"A + B = \n{AB_add}")
print(f"A - B = \n{AB_sub}")
print(f"A . B = \n{AB_mul}")

print("\nc. Tính phép nhân ma trận Y = A x X")
print(f"Y = A x X --> shape A ({A.shape}) x shape X ({X.shape}) = shape Y ({Y.shape}) \n{Y}")
```

#### 4. Giải hệ phương trình tuyến tính
Giải hệ phương trình tuyến tính sau:
$$
\left\{\begin{matrix}
        4x_1  + 3x_2 - 5x_3 = 2 \\ 
        -2x_1 - 4x_2 + 5x_3 = 5 \\ 
        8x_1 + 8x_2 = -3 \\ 
\end{matrix}\right.
$$

```python
# Tạo ma trận A (ví dụ 3x3)
A = np.array([
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
], dtype=float)

# Tạo vector y
y = np.array([8, -11, -3], dtype=float)

# Giải hệ Ax = y
x = np.linalg.solve(A, y)

# Kiểm tra lại: yy = A @ x
yy = A @ x

# Tính chuẩn L2 của sai số
diff = np.linalg.norm(yy - y, 2)

print(f"+ Ma tran A: \n{A}")
print(f"+ Vector y: {y}")
print(f"+ Nghiem x: {x}")
print(f"+ Kiem tra: yy = Ax = {yy} ==> ||yy -y ||_2 = {diff: .2f}")
```

### Bài 3. Tính gần đúng số pi dùng phương pháp Monte Carlo

Vẽ một đường tròn $C$ có bán kính $r$ nội tiếp một hình vuông $C$, và đặt ngẫu nhiên $N$ chấm lên hình vuông. Tỉ lệ các chấm nằm trong hình tròn trên tổng số chấm xấp xỉ bằng diện tích của hình tròn chia cho hình vuông như sau:
$$
\frac{N_C}{N} \approx  \frac{\mathbb{S}_C}{\mathbb{S}_N}
$$
nơi mà $N_C$ là các chấm trong hình tròn, $N$ là tổng số chấm, $\mathbb{S}_C$ và $\mathbb{S}_N$ lần lượt là diện tích của hình tròn và hình vuông.
$$ \mathbb{S}_C = \pi * r^2 $$
$$ \mathbb{S}_N = (2*r)^2 = 4r^2 $$
Do đó, công thức trên trở thành:
$$
\frac{N_C}{N} \approx \frac{\pi * r^2}{4r^2} = \frac{\pi}{4}
$$
Suy ra, ta có giá trị số $\pi$ được tính như sau:
$$
\pi \approx \frac{4N_C}{N}
$$

Các bạn hãy lập trình tính sấp xỉ số $\pi$ dùng phương pháp Monte Carlo. Cho biết các sai số với $N = 100$, $N = 10000$, và $N = 1000000$. Vẽ đồ thị minh họa cho chương trình.

**Hướng dẫn**: Viết chương trình phát sinh $N$ điểm $(x_i, y_i)$ với $x_i, y_i \in [-1, -1]$. $(x_i, y_i)$ thuộc hình tròn tâm $(0, 0)$ bán kính 1 nội tiếp hình vuông có tâm tại $(0, 0)$ độ dài cạnh 2 khi và chỉ khi $x_i^2 + y_i^2 <= 1$.

```python
import matplotlib.pyplot as plt # khai báo thư viện vẽ pyplot
import math

circle = plt.Circle((0, 0), 1, color='g') # tạo đối tượng vòng tròn

fig, ax = plt.subplots(figsize=(6, 6)) # lấy figure và vùng vẽ

plt.plot(0, 0, 'o', color=(0.9, 0.9, 1.0), alpha=0.8) # vẽ điểm tại tâm (0,0)
ax.add_patch(circle) # thêm vòng tròn vào vùng vẽ

x = 0.75; y = math.sqrt(1 - (x ** 2)) # (x = 0.75, y = căn bậc 2 của 1 - x^2)
plt.arrow(0, 0, x, y)

plt.xlim(-1, 1) # giới hạn trục x
plt.ylim(-1, 1) # giới hạn trục y

plt.savefig('pi.png') # lưu đồ thị xuống tập tin
plt.show()            # hiển thị đồ thị
```

```python
import numpy as np

def calc_pi_monte_carlo(n = 100):

    x = np.random.rand(n)
    y = np.random.rand(n)
    
    # Kiểm tra điểm nào nằm trong 1/4 hình tròn bán kính 1
    inside = x**2 + y**2 <= 1
    
    # Ước lượng pi
    pi = 4 * np.sum(inside) / n
    
    return pi
# calc_pi_monte_carlo
```

```python
import math

print("epsilon(n=100): ", calc_pi_monte_carlo(n = 100) - math.pi)
print("epsilon(n=100): ", calc_pi_monte_carlo(n = 100) - math.pi)
print("epsilon(n=10000): ", calc_pi_monte_carlo(n = 10000) - math.pi)
print("epsilon(n=10000): ", calc_pi_monte_carlo(n = 10000) - math.pi)
print("epsilon(n=1000000): ", calc_pi_monte_carlo(n = 1000000) - math.pi)
print("epsilon(n=1000000): ", calc_pi_monte_carlo(n = 1000000) - math.pi)
```
